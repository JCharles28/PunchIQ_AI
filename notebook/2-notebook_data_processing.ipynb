```python
import pandas as pd
```

## Load data

```python
# Load the dataset
dataset = pd.read_csv('../data/fighters.csv')
dataset.head()
```

## Clean data

- Handle columns with 'Unknown' values  
    + Scrape Wikipedia to remove each 'Unknown' value  
- Convert the 'ko_rate' column to float  
- Convert the height to a format (**cm** preferred)
- Modify the 'Name' column so that each word starts with a capital letter

```python
# Default data
dataset = dataset.dropna()  # Drop rows with missing values
dataset = dataset.drop_duplicates()  # Drop duplicate rows

# drop lines with 'Unknown' values in every column
dataset = dataset[~dataset.isin(['Unknown']).any(axis=1)]

# add columns for height_cm , reach_cm
dataset['height_cm'] = dataset['height'].apply(lambda x: float(x.split()[0]) * 30.48 if 'ft' in x else float(x.split()[0]))
dataset['reach_cm'] = dataset['reach'].apply(lambda x: float(x.split()[0]) * 2.54 if 'in' in x else float(x.split()[0]))

# drop columns that are not needed: height, reach
dataset = dataset.drop(columns=['height', 'reach'])

# change 'ko_rate' : percentage to float
dataset['ko_rate'] = dataset['ko_rate'].str.rstrip('%').astype(float) / 100.0

# change 'name' to lowercase
dataset['name'] = dataset['name'].str.lower()
```

## Save data

```python
dataset.to_csv('../data/fighters_cleaned.csv', index=False)
```
